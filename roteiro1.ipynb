```python
# Carregando as bibliotecas solicitadas no roteiro
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload da imagem
print("Selecione uma imagem colorida (JPG ou PNG):")
uploaded = files.upload()

# Lendo a imagem no Colab
nome_arquivo = list(uploaded.keys())[0]
nparr = np.frombuffer(uploaded[nome_arquivo], np.uint8)
img = cv2.imdecode(nparr, cv2.IMREAD_COLOR)

# Exibindo a imagem em BGR
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Visualização Incorreta (BGR)")

# Corrigindo a imagem de BGR para RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.subplot(1, 2, 2)
plt.imshow(img_rgb)
plt.title("Visualização Correta (RGB)")
plt.show()
```

```python
# Fazendo uma cópia para não alterar a imagem original
img_ex2 = img.copy()

# Desenhando um quadrado preto no canto superior esquerdo (50x50)
img_ex2[0:50, 0:50] = (0, 0, 0)

# Canal Vermelho no centro da imagem
altura, largura, canais = img_ex2.shape
centro_y = altura // 2
centro_x = largura // 2

# Criando uma área 100x100 no centro e alterando apenas o canal Vermelho (índice 2) para 255
img_ex2[centro_y-50:centro_y+50, centro_x-50:centro_x+50, 2] = 255

# Exibindo o resultado final
plt.imshow(cv2.cvtColor(img_ex2, cv2.COLOR_BGR2RGB))
plt.title("Exercício 2 - Manipulação de Matriz")
plt.show()
```

```python
# Converter a imagem original colorida para Tons de Cinza
img_cinza = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Criar uma nova imagem toda preta com o mesmo tamanho da imagem cinza
img_pb = np.zeros_like(img_cinza)

# Lógica de Binarização:
img_pb[img_cinza > 127] = 255

# Resultado final
plt.imshow(img_pb, cmap='gray')
plt.title("Exercício 3 - Imagem Binarizada")
plt.show()
```
